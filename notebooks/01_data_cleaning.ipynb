```python
import sqlite3
import pandas as pd
import os

# Root folder mein maujud database ka relative path
db_path = '../zomato_project.db'

conn = sqlite3.connect(db_path)
print("Database connected successfully!")

# Database ki sabhi 12 tables ke naam fetch karna
tables_query = "SELECT name FROM sqlite_master WHERE type='table';"
tables = pd.read_sql(tables_query, conn)
print("Available tables:", tables['name'].tolist())

# Sabhi tables ko ek dictionary mein load karna aur unka shape check karna
dfs = {}
for table in tables['name']:
    dfs[table] = pd.read_sql(f"SELECT * FROM {table}", conn)
    print(f"Loaded {table}: {dfs[table].shape}")
```

```python
# Har table ke liye duplicates aur missing values check karna
for name, df in dfs.items():
    print(f"\n--- Processing Table: {name} ---")
    
    # Duplicate rows hatana
    initial_rows = len(df)
    df.drop_duplicates(inplace=True)
    print(f"Dropped {initial_rows - len(df)} duplicate rows.")
    
    # Missing values check karna
    missing = df.isnull().sum()
    missing_cols = missing[missing > 0]
    if not missing_cols.empty:
        print("Missing values found:")
        print(missing_cols)
    else:
        print("No missing values.")
```

```python
# Processed folder create karna agar pehle se na ho
os.makedirs('../data/processed', exist_ok=True)

# Sabhi cleaned tables ko CSV format mein save karna
for name, df in dfs.items():
    output_path = f'../data/processed/cleaned_{name}.csv'
    df.to_csv(output_path, index=False)
    print(f"Saved cleaned_{name}.csv")

# Connection close karna
conn.close()
print("\nData cleaning and export process completed successfully!")
```
